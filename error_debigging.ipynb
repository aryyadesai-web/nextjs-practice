An exception is an error condition that occurs during program execution and can be handled using exception-handling mechanisms.

```python
#try and except 
try:
    a=10
    b=0
    print(a/b)
except ZeroDivisionError:
    print("division by zero is not possible")
```

```python
#try and except 
try:
    a=10
    b=0
    result=a/b
    print(result)
except ZeroDivisionError:
    print("division by zero is not possible")   
```
