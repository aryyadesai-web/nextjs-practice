```python
import matplotlib.pyplot as plt
months=["jan","feb","mar","apr","may"]
python=[100,150,200,180,200]
sql=[50,100,150,120,130]
powerbi=[20,30,40,50,60]
plt.plot(months,python,label="Python")
plt.plot(months,sql,label="SQL")
plt.plot(months,powerbi,label="PowerBI")
plt.title("Multiple Line Charts")
plt.xlabel("Months")
plt.ylabel("Sales") 
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
students=["A","B","C","D"]
marks=[50,40,30,20]
bars=plt.bar(students,marks,)
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        bar.get_height(),
        ha='center'
    )
plt.show()
```

```python
import matplotlib.pyplot as plt
height=[150,125,225,155,100]
weight=[10,25,30,56,45]
plt.scatter(height,weight,color="blue",s=200,alpha=0.5,marker="*")
plt.title("Scatter chart")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.show()
```
