```python
def greet(name):
    print("HHello " + name)
greet("Arya")

#name is the parameter and "Arya" is the argument. The parameter is a variable that is used in the function definition, while the argument is the actual value that is passed to the function when it is called.
```

```python
def add(a,b):
    print(a+b)
result = add(5,6)    
print(result)    
```

```python
def add(a,b):
    return a+b
result = add(5,6)   
print(result)
#print displays information on the screen while return sends information back to the caller of the function. When you use print, the function will display the result on the screen, but it won't give you a value that you can use later in your code. When you use return, the function will give you a value that you can store in a variable or use in other calculations.  
```

```python
#types of arguments 
#positional arguments: Values are assigned according to their positions
def stud(name,age):
    print(name,age)
stud("Arya",20)
```

```python
#keyword arguments: Values are assigned using parameter names
def stud(name,age):
    print(name,age)
stud(age=20,name="Arya")
```

```python
#default arguments: A parameter has predefined value if the caller does not provide a value for that parameter
def greet(name,msg="hi"):
    print(msg,name)
greet("Arya")
print("arya","desai")
```

```python
#*args: used when you don't know how many positional arguments will be passed
def calc(*num):
    print(num)
    print(sum(num))
calc(10,20,30)
```

```python
#**kwargs: used when you don't know how many keyword arguments will be passed
def display(**details):
    print(details)
display(name="Arya",age=20,city="Mumbai")
```

```python
#Variable scope
#local variable: created inside the function
def dislpay():
    msg="arya"
    print(msg)
dislpay()
```

```python
#global variable: created outside the function
name="arya"
def display():
    print(name)
display()
```

```python
#using global keyword: used to modify a global variable inside a function
count=10
def update():
    global count
    count = 20
update()
print(count)    


```

```python
#lambda functions 
sq=lambda x:x*x
print(sq(4))
```

```python
add=lambda a,b:a+b
print(add(7,8))
```

```python
#recursion: A function that calls itself is called a recursive function. Recursion is a programming technique where a function calls itself in order to solve a problem. It is often used to solve problems that can be broken down into smaller subproblems of the same type.
def fact(n):
    if n==1:
        return 1
    return n*fact(n-1)
print(fact(5))
```

```python
#module: a module is a file containing Python definitions and statements. The file name is the module name with the suffix .py added. A module can define functions, classes, and variables. A module can also include runnable code.   
def add(a,b):
    return a+b
def sub(c,d):   
     return c-d
print(add(5,6))
print(sub(10,4))    
```
