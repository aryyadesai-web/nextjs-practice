A customer support center receives an average of of 4 calls per min.
Assuming the number of calls follows a poission distribution calc:
1. The proablitlity of receing exactly 2 calls in a min
2. The proablitlity of receing at most 2 calls in a min
3. The proablitlity of receing more than 2 calls in a min

```python

from scipy.stats import poisson
lam = 4
print("Exactly two calls:", poisson.pmf(2, lam))
print("At most two calls:", poisson.cdf(2, lam))
print("More than two calls:", poisson.sf(2, lam))
```

A company finds that proability of customer purchasing a product is 0.6.
If 5 customers are selected independently, calcuate.
1. The probality of exactly 3 customers purchase the product 
2. The probality of at most 3 customers purchase the product 
3. The probality of more than 3 customers purchase the product 

```python
from scipy.stats import binom
n = 5
p = 0.6
print("Exactly three products:", binom.pmf(3, n, p))
print("At most three products:", binom.cdf(3, n, p))
print("More than three products:", binom.sf(3, n, p))
```

Continous distribution 
Normal distribtuion with z score
Package dilvery time has mean 40 mins and SD 50 mins.
What propotion takes under 47 mins

```python
from scipy.stats import norm
mu, sigma, x = 40,5,47
z=(x-mu)/sigma
p_under = norm.cdf(x, loc=mu, scale=sigma)
print("Z-Score: ", round(z,2))
print("P(time<47):", round(p_under,4))
```

Company claims average packing time is 30 mins 
H0: Average packing time = 30 mins
H1: Average packign time != 30 mins 

```python
#Test whether avg packing time differs from 30 mins 
import numpy as np
from scipy import stats
times = np.array([32, 29, 31, 30, 28, 33, 27, 34, 30, 29])
t_stat, p_value = stats.ttest_1samp(times,popmean= 30)
print("Mean:", round(times.mean(),2))
print("T-statistic:", round(t_stat,4))
print("P-value:", round(p_value,4))
```
